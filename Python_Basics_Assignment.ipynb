**Question 2 : Explain the difference between List and Tuple in Python.**

Answer:- Both List and Tuple are used to store multiple values in Python, but they have some important differences:

| Feature         | List                            | Tuple                         |
| --------------- | ------------------------------- | ----------------------------- |
| **Syntax**      | `[ ]`                           | `( )`                         |
| **Mutability**  | Mutable – can be changed        | Immutable – cannot be changed |
| **Performance** | Relatively slower               | Relatively faster             |
| **Use**         | When data may need modification | When data should remain fixed |


**Question 3 : What is a function in Python? Why are functions useful?**

Answer:- A function in Python is a reusable block of code that performs a specific task. It is defined using the def keyword.

**Why are functions useful**?

**Code Reusability**: A function can be used multiple times without writing the same code again.

**Reduces Complexity**: Functions divide a large program into smaller, manageable parts.

**Saves Time**: Once created, a function can be called whenever needed.
Easy Maintenance: Changes can be made in one place instead of modifying repeated code.

In short: Functions make Python programs organized, reusable, and easier to understand and maintain.

**Question 5 : Write a Python program to check whether a number is even or odd using conditional statements.**

```python
#Answer:-
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is Even")
else:
    print("The number is Odd")
```

**Question 6 : Write a program to print numbers from 1 to 10 using a loop.**

**Question 7 : Create a list of five numbers and print the maximum number from the list.**

```python
#Answer:-
list_1= [10,20,30,40,50]
max_num = list_1[0]

for num in list_1:
    if num > max_num:
        max_num = num

print("The maximum number in the list is:", max_num)
```

**Question 10 : Write a Python program to count how many times a number appears in a list.** **Example List:- [2,3,4,2,5,2]**

```python
#Answer:-
numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print("The number", num, "appears", count, "times.")
```
